# SCR Analysis: Acquisition, Extinction, And Reinstatement

This production-ready notebook analyzes trial-wise skin conductance response (SCR) across acquisition, extinction, and reinstatement.

SCR scoring follows the project rule:

- Raw SCR amplitude is `phaBase2Peak`.
- Negative values and values below `0.02 uS` are coded as exactly zero.
- The analysis score is `sqrt_scr = sqrt(thresholded SCR)`.
- Primary analyses retain all participants.
- SCR-defined responders and learners are sensitivity subsets only.

Learner definitions match `identify_fear_learning_subjects_scr.ipynb`:

- Physiological SCR responder: at least two acquisition CS+ trials with raw SCR >= `0.05 uS`.
- Simple acquisition differential learner: physiological responder with mean CS+ > CS- across acquisition.
- Habituation-adjusted learner: physiological responder with a positive CS+ coefficient from `sqrtSCR ~ CSplus + Trial_Z`.
- Late-phase sensitivity learner: late CS+ > CS- and late discrimination greater than early discrimination.

Statistical plan implemented here:

- Acquisition: fear learning is tested as CS+ vs CS-, with SAD vs HC comparison.
- Extinction: vicarious fear extinction is tested primarily as CSS vs CSR, with CSS vs CS-, CSR vs CS-, and `(CSS - CS-) - (CSR - CS-)` also reported.
- Reinstatement: vicarious fear reinstatement uses the same targeted topology as extinction.
- Extinction/reinstatement between-group inference includes Group x Drug interaction tests.
- Dynamic models quantify trial-wise trajectories using mixed-effects models where possible.


## Analysis Proposal: CSS/CSR Dynamics And SAD-HC Differences

### Core question

This SCR analysis is designed to test whether SAD and HC differ in how they physiologically update responses to `CSS` and `CSR` across extinction and reinstatement. The key question is not only whether `CSR > CSS` on average, but whether the `CSR - CSS` trajectory changes differently across trials and phases in SAD vs HC.

### Primary SCR score

Trial-wise SCR amplitudes are scored from raw `phaBase2Peak`. Negative values and values below `0.02 uS` are coded as zero, then square-root transformed. Primary neural/behavioral analyses retain all participants; SCR-defined responder/learner groups are sensitivity analyses.

### Primary contrasts

The phase-level contrasts are organized as:

- Acquisition: `CS+ - CS-`
- Extinction: `CSR - CSS`, `CSS - CS-`, `CSR - CS-`
- Reinstatement: `CSR - CSS`, `CSS - CS-`, `CSR - CS-`

The main theoretical contrast is `CSR - CSS`, because it directly tests whether threat-like responding to CSR exceeds safety-like responding to CSS.

### Primary dynamic model

For extinction and reinstatement separately, the primary inferential model should be:

```text
sqrtSCR ~ Condition(CSR vs CSS) * Trial_Z * Group + Drug + (1 | Subject)
```

Primary terms:

- `Condition[CSR]`: average CSR-CSS difference.
- `Condition[CSR] x Trial_Z`: whether CSR-CSS changes across trials.
- `Condition[CSR] x Group[SAD]`: whether SAD differs from HC in average CSR-CSS.
- `Condition[CSR] x Trial_Z x Group[SAD]`: whether SAD differs from HC in the dynamic CSR-CSS trajectory.

If phase comparison is the goal, use the integrated model:

```text
sqrtSCR ~ Phase * Condition(CSR vs CSS) * Trial_Z * Group + Drug + (1 | Subject)
```

The key term is `Phase x Condition x Trial_Z x Group`, which tests whether SAD-HC differences in CSR-CSS dynamics differ between extinction and reinstatement.

### Visual and sensitivity strategy

Use the sliding-window figures as descriptive localization of effects over time:

- 3-trial windows: `T1-T3` through `T6-T8`
- 2-trial windows: `T1-T2` through `T7-T8`
- Trial-1-excluded windows: `All = T2-T8`, `Early = T2-T4`, `Late = T5-T8`

Each window is plotted for all subjects, each diagnostic group, and each `Group x Drug` cell. Stars on these plots mark uncorrected one-sample tests where the displayed contrast is greater than zero. These are descriptive markers, not corrected confirmatory claims.

Primary inference should come from the mixed-effects dynamic models; the sliding-window plots explain when and in which sensitivity sets effects appear.

### Working hypothesis

SAD participants may show greater persistence of threat-related physiological differentiation, reflected by stronger or less declining `CSR - CSS` during extinction and/or reinstatement compared with HC. In reinstatement, SAD may show stronger re-emergence or maintenance of `CSR > CSS`, consistent with reduced safety generalization or impaired regulation of vicariously acquired threat representations.

````python
from pathlib import Path
import math
import warnings

import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.formula.api as smf
from IPython.display import display, Markdown

pd.set_option("display.max_rows", 240)
pd.set_option("display.max_columns", 220)
warnings.filterwarnings("ignore", category=RuntimeWarning)

DATA_DIR = Path("/Users/xiaoqianxiao/projects/NARSAD/EDR")
ACQ_PATH = DATA_DIR / "peak_stats_table-phase1.csv"
PHASE23_PATH = DATA_DIR / "peak_stats_table-phase2.3.csv"
META_PATH = Path("/Users/xiaoqianxiao/projects/NARSAD/MRI/source_data/behav/drug_order.csv")
OUT_DIR = Path("scr_analysis_outputs")
OUT_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR = OUT_DIR / "figures_publication"
FIG_DIR.mkdir(parents=True, exist_ok=True)

RAW_COL = "phaBase2Peak"
THRESH_COL = "phaBase2Peak_thresholded_0p02"
SCORE_COL = "sqrt_phaBase2Peak"
BAD_COL = "bad"
RESPONSE_THRESHOLD_RAW = 0.02
PHYSIO_RESPONDER_THRESHOLD_RAW = 0.05
PHYSIO_RESPONDER_MIN_CSPLUS_TRIALS = 2
WINDOW_N = 4

CONDITION_ORDER = ["CS-", "CSS", "CSR"]
DOMAIN_ORDER = ["CS-", "CS+"]
PHASE_ORDER = ["Acquisition", "Extinction", "Reinstatement"]
GROUP_ORDER = ["HC", "SAD"]
ANALYSIS_SET_ORDER = [
    "All subjects",
    "Physiological responders",
    "Simple CS+ learners",
    "Habituation-adjusted CS+ learners",
    "Late-phase sensitivity CS+ learners",
]
CONTRAST_ORDER = ["CS+ - CS-", "CSS - CS-", "CSR - CS-"]
PHASE_METRIC_ORDER = ["CS+ - CS-", "CSR - CSS", "CSS - CS-", "CSR - CS-", "(CSR - CS-) - (CSS - CS-)"]

GROUP_PALETTE = {"HC": "#0072B2", "SAD": "#D55E00"}
CONDITION_PALETTE = {"CS-": "#0072B2", "CSS": "#D55E00", "CSR": "#009E73", "CS+": "#CC79A7"}
SET_PALETTE = {
    "All subjects": "#3B6EA8",
    "Physiological responders": "#6A51A3",
    "Simple CS+ learners": "#2F9E44",
    "Habituation-adjusted CS+ learners": "#C44E52",
    "Late-phase sensitivity CS+ learners": "#DD8452",
}

sns.set_theme(style="ticks", context="paper", font_scale=1.18)
plt.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 300,
    "font.family": "Arial",
    "axes.linewidth": 1.0,
    "axes.edgecolor": "#222222",
    "axes.titleweight": "regular",
    "axes.labelsize": 10,
    "axes.titlesize": 11,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 8.5,
    "legend.title_fontsize": 9,
})

print("ACQ_PATH:", ACQ_PATH, "exists=", ACQ_PATH.exists())
print("PHASE23_PATH:", PHASE23_PATH, "exists=", PHASE23_PATH.exists())
print("META_PATH:", META_PATH, "exists=", META_PATH.exists())
print("OUT_DIR:", OUT_DIR.resolve())


def save_current_fig(name, dpi=300):
    png_path = FIG_DIR / name
    pdf_path = FIG_DIR / name.replace(".png", ".pdf")
    plt.savefig(png_path, dpi=dpi, bbox_inches="tight", facecolor="white")
    plt.savefig(pdf_path, bbox_inches="tight", facecolor="white")
    print("Saved figure ->", png_path)
    return png_path


def safe_percent(n, d):
    return 100 * n / d if d else np.nan


def p_to_stars(p):
    if pd.isna(p):
        return ""
    if p < 0.001:
        return "***"
    if p < 0.01:
        return "**"
    if p < 0.05:
        return "*"
    return ""


def p_text(p):
    if pd.isna(p):
        return "p=NA"
    if p < 0.001:
        return "p<.001"
    return f"p={p:.3f}".replace("0.", ".")


def safe_mean(series):
    x = pd.to_numeric(pd.Series(series), errors="coerce").dropna()
    return float(x.mean()) if len(x) else np.nan


def clean_axis(ax):
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#222222")
    ax.spines["bottom"].set_color("#222222")
    ax.grid(True, axis="y", color="#E6E6E6", linewidth=0.7)
    ax.grid(False, axis="x")
    ax.tick_params(axis="both", length=3, width=0.8, color="#222222")
    return ax


def panel_label(ax, label):
    ax.text(-0.12, 1.06, label, transform=ax.transAxes, fontsize=12, fontweight="bold", va="top", ha="left")


def metric_display_name(metric):
    labels = {
        "CS+ - CS-": "CS+ - CS-",
        "CSR - CSS": "CSR - CSS",
        "CSS - CS-": "CSS - CS-",
        "CSR - CS-": "CSR - CS-",
        "(CSR - CS-) - (CSS - CS-)": "ΔCSR - ΔCSS",
    }
    return labels.get(metric, metric)

````

## 2. Load, Score, And Annotate SCR


````python
def load_metadata(path=META_PATH):
    if not path.exists():
        print(f"Metadata not found: {path}. Group comparisons will be unavailable.")
        return pd.DataFrame(columns=["Subject", "Group", "Drug"])
    meta = pd.read_csv(path)
    if "subject_id" not in meta.columns:
        raise ValueError(f"Metadata file {path} does not contain subject_id")
    meta = meta.copy()
    meta["Subject"] = meta["subject_id"].astype(str).str.strip()
    keep = [c for c in ["Subject", "Group", "Drug", "drug_condition", "Gender", "gender_code", "demo_age", "guess"] if c in meta.columns]
    meta = meta[keep].drop_duplicates("Subject")
    if "Group" in meta.columns:
        meta["Group"] = meta["Group"].astype(str).str.strip()
    if "Drug" in meta.columns:
        meta["Drug"] = meta["Drug"].astype(str).str.strip()
    return meta


def load_scr_file(path, phase_name=None):
    df = pd.read_csv(path)
    required = {"sid", "stTy", "stNum", RAW_COL}
    missing = sorted(required - set(df.columns))
    if missing:
        raise ValueError(f"{path} is missing required columns: {missing}")
    df = df.copy()
    df["Subject"] = df["sid"].astype(str).str.strip()
    df["stNum"] = pd.to_numeric(df["stNum"], errors="coerce")
    df[RAW_COL] = pd.to_numeric(df[RAW_COL], errors="coerce")
    if BAD_COL in df.columns:
        df[BAD_COL] = pd.to_numeric(df[BAD_COL], errors="coerce").fillna(0).astype(int)
        df = df[df[BAD_COL] == 0].copy()
    df["Condition"] = df["stTy"].replace({"CS+S": "CSS", "CS+R": "CSR", "CS-": "CS-"})
    df = df[df["Condition"].isin(CONDITION_ORDER)].copy()
    df["Domain"] = np.where(df["Condition"].isin(["CSS", "CSR"]), "CS+", "CS-")
    df[THRESH_COL] = np.where(df[RAW_COL] >= RESPONSE_THRESHOLD_RAW, df[RAW_COL], 0.0)
    df[SCORE_COL] = np.sqrt(df[THRESH_COL])
    if phase_name is not None:
        df["Phase"] = phase_name
    df = df.dropna(subset=["Subject", "stNum", "Condition", RAW_COL, THRESH_COL, SCORE_COL])
    return df

meta = load_metadata()
acq = load_scr_file(ACQ_PATH, phase_name="Acquisition")
phase23 = load_scr_file(PHASE23_PATH)
extinction = phase23[(phase23["stNum"] >= 1) & (phase23["stNum"] <= 24)].copy()
extinction["Phase"] = "Extinction"
reinstatement = phase23[(phase23["stNum"] >= 25) & (phase23["stNum"] <= 48)].copy()
reinstatement["Phase"] = "Reinstatement"

all_scr = pd.concat([acq, extinction, reinstatement], ignore_index=True)
if not meta.empty:
    all_scr = all_scr.merge(meta, on="Subject", how="left")
else:
    all_scr["Group"] = np.nan
    all_scr["Drug"] = np.nan

all_scr["Phase"] = pd.Categorical(all_scr["Phase"], categories=PHASE_ORDER, ordered=True)
all_scr["Condition"] = pd.Categorical(all_scr["Condition"], categories=CONDITION_ORDER, ordered=True)
all_scr["ConditionTrial"] = all_scr.groupby(["Subject", "Phase", "Condition"], observed=False).cumcount() + 1
all_scr["PhaseTrial"] = all_scr.groupby(["Subject", "Phase"], observed=False).cumcount() + 1
all_scr["Window"] = np.where(all_scr["ConditionTrial"] <= WINDOW_N, "Early", np.where(all_scr["ConditionTrial"] > 8 - WINDOW_N, "Late", "Middle"))
all_scr["Trial_Z"] = all_scr.groupby(["Subject", "Phase"], observed=False)["PhaseTrial"].transform(lambda x: (x - x.mean()) / x.std(ddof=0) if x.std(ddof=0) > 0 else 0)

print("Rows and subjects by phase")
display(all_scr.groupby("Phase", observed=False).agg(rows=("Subject", "size"), subjects=("Subject", "nunique")))
print("Trial count by phase and condition")
display(all_scr.groupby(["Phase", "Condition"], observed=False).size().unstack(fill_value=0))
print("Subjects by Group x Drug")
display(all_scr[["Subject", "Group", "Drug"]].drop_duplicates().groupby(["Group", "Drug"], dropna=False).size().rename("subjects").reset_index())
print("SCR distribution after scoring")
display(all_scr[[RAW_COL, THRESH_COL, SCORE_COL]].describe())
````

````python
floor_rows = []
for phase in PHASE_ORDER:
    sub = all_scr[all_scr["Phase"] == phase]
    for condition in CONDITION_ORDER:
        g = sub[sub["Condition"] == condition]
        floor_rows.append({
            "Phase": phase,
            "Condition": condition,
            "N_trials": int(len(g)),
            "N_subjects": int(g["Subject"].nunique()),
            "Raw_mean": float(g[RAW_COL].mean()),
            "Raw_median": float(g[RAW_COL].median()),
            "Sqrt_mean": float(g[SCORE_COL].mean()),
            "Sqrt_median": float(g[SCORE_COL].median()),
            "Percent_raw_below_0p02": float((g[RAW_COL] < RESPONSE_THRESHOLD_RAW).mean() * 100),
            "Percent_raw_ge_0p05": float((g[RAW_COL] >= PHYSIO_RESPONDER_THRESHOLD_RAW).mean() * 100),
        })
floor_summary = pd.DataFrame(floor_rows)
display(floor_summary)

floor_subject_summary = (
    all_scr.groupby(["Subject", "Group", "Phase", "Condition"], observed=True)
    .agg(
        Sqrt_mean=(SCORE_COL, "mean"),
        Percent_raw_below_0p02=(RAW_COL, lambda s: float((s < RESPONSE_THRESHOLD_RAW).mean() * 100)),
        N_trials=(RAW_COL, "size"),
    )
    .reset_index()
)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8))
sns.barplot(
    data=floor_subject_summary, x="Phase", y="Sqrt_mean", hue="Condition",
    order=PHASE_ORDER, hue_order=CONDITION_ORDER, palette=CONDITION_PALETTE,
    errorbar="se", capsize=0.08, err_kws={"linewidth": 1.1}, ax=axes[0]
)
axes[0].set_title("Mean thresholded sqrt SCR")
axes[0].set_ylabel("sqrt SCR")
axes[0].set_xlabel("")
clean_axis(axes[0])
sns.barplot(
    data=floor_subject_summary, x="Phase", y="Percent_raw_below_0p02", hue="Condition",
    order=PHASE_ORDER, hue_order=CONDITION_ORDER, palette=CONDITION_PALETTE,
    errorbar="se", capsize=0.08, err_kws={"linewidth": 1.1}, ax=axes[1]
)
axes[1].set_title("Floor-rate after 0.02 uS threshold")
axes[1].set_ylabel("Trials below 0.02 uS (%)")
axes[1].set_xlabel("")
clean_axis(axes[1])
axes[1].legend_.remove()
plt.tight_layout()
save_current_fig("scr_description_mean_and_floor_by_phase.png")
plt.show()

# Acquisition-specific habituation evidence: early and late bars, not just difference.
hab_rows = []
for condition in CONDITION_ORDER:
    for subject, g in all_scr[(all_scr["Phase"] == "Acquisition") & (all_scr["Condition"] == condition)].groupby("Subject"):
        for window in ["Early", "Late"]:
            gw = g[g["Window"] == window]
            hab_rows.append({
                "Subject": subject,
                "Group": gw["Group"].dropna().iloc[0] if gw["Group"].notna().any() else np.nan,
                "Condition": condition,
                "Window": window,
                "Mean_sqrtSCR": safe_mean(gw[SCORE_COL]),
                "Floor_percent": float((gw[RAW_COL] < RESPONSE_THRESHOLD_RAW).mean() * 100) if len(gw) else np.nan,
            })
hab_df = pd.DataFrame(hab_rows)

hab_stats = []
for condition in CONDITION_ORDER:
    wide = hab_df[hab_df["Condition"] == condition].pivot(index="Subject", columns="Window", values="Mean_sqrtSCR").dropna()
    diff = wide["Late"] - wide["Early"]
    t = stats.ttest_rel(wide["Late"], wide["Early"], alternative="less") if len(diff) >= 2 else None
    wide_floor = hab_df[hab_df["Condition"] == condition].pivot(index="Subject", columns="Window", values="Floor_percent").dropna()
    fdiff = wide_floor["Late"] - wide_floor["Early"]
    ft = stats.ttest_rel(wide_floor["Late"], wide_floor["Early"], alternative="greater") if len(fdiff) >= 2 else None
    hab_stats.append({
        "Condition": condition,
        "N": len(diff),
        "Early_sqrt_mean": float(wide["Early"].mean()),
        "Late_sqrt_mean": float(wide["Late"].mean()),
        "Late_minus_Early_sqrt": float(diff.mean()),
        "t_late_less_early": float(t.statistic) if t is not None else np.nan,
        "p_late_less_early": float(t.pvalue) if t is not None else np.nan,
        "Early_floor_percent": float(wide_floor["Early"].mean()),
        "Late_floor_percent": float(wide_floor["Late"].mean()),
        "Late_minus_Early_floor_percent": float(fdiff.mean()),
        "t_floor_late_greater_early": float(ft.statistic) if ft is not None else np.nan,
        "p_floor_late_greater_early": float(ft.pvalue) if ft is not None else np.nan,
    })
habituation_stats = pd.DataFrame(hab_stats)
print("Acquisition habituation/floor paired statistics")
display(habituation_stats)

fig, axes = plt.subplots(1, 2, figsize=(8.8, 3.8))
window_palette = {"Early": "#0072B2", "Late": "#D55E00"}
sns.barplot(data=hab_df, x="Condition", y="Mean_sqrtSCR", hue="Window", order=CONDITION_ORDER, hue_order=["Early", "Late"], palette=window_palette, errorbar="se", capsize=0.10, err_kws={"linewidth": 1.1}, ax=axes[0])
axes[0].set_title("Habituation across acquisition")
axes[0].set_ylabel("Mean SCR (sqrt uS)")
axes[0].set_xlabel("")
axes[0].set_ylim(0, 0.24)
clean_axis(axes[0])
p_y = 0.225
for idx, row in habituation_stats.iterrows():
    axes[0].text(idx, p_y, p_text(row["p_late_less_early"]), ha="center", va="bottom", fontsize=7.5)
sns.barplot(data=hab_df, x="Condition", y="Floor_percent", hue="Window", order=CONDITION_ORDER, hue_order=["Early", "Late"], palette=window_palette, errorbar="se", capsize=0.10, err_kws={"linewidth": 1.1}, ax=axes[1])
axes[1].set_title("Floor effect across acquisition")
axes[1].set_ylabel("Trials below 0.02 uS (%)")
axes[1].set_xlabel("")
axes[1].set_ylim(0, 92)
clean_axis(axes[1])
axes[1].legend_.remove()
for idx, row in habituation_stats.iterrows():
    axes[1].text(idx, 87, p_text(row["p_floor_late_greater_early"]), ha="center", va="bottom", fontsize=7.5)
handles, labels = axes[0].get_legend_handles_labels()
axes[0].legend(handles[:2], labels[:2], title="Window", frameon=False, loc="upper right")
plt.tight_layout()
save_current_fig("scr_acquisition_habituation_floor_early_late_bars.png")
plt.show()

````

## 4. Define Responders And Learners


````python
def build_physio_flags(acq_df):
    rows = []
    for subject, g in acq_df.groupby("Subject"):
        csplus = g[g["Domain"] == "CS+"]
        css = g[g["Condition"] == "CSS"]
        csr = g[g["Condition"] == "CSR"]
        rows.append({
            "Subject": subject,
            "n_raw_CSplus_ge_0p05": int((csplus[RAW_COL] >= PHYSIO_RESPONDER_THRESHOLD_RAW).sum()),
            "n_raw_CSS_ge_0p05": int((css[RAW_COL] >= PHYSIO_RESPONDER_THRESHOLD_RAW).sum()),
            "n_raw_CSR_ge_0p05": int((csr[RAW_COL] >= PHYSIO_RESPONDER_THRESHOLD_RAW).sum()),
        })
    out = pd.DataFrame(rows)
    out["physiological_responder_CSplus"] = out["n_raw_CSplus_ge_0p05"] >= PHYSIO_RESPONDER_MIN_CSPLUS_TRIALS
    out["physiological_responder_CSS"] = out["n_raw_CSS_ge_0p05"] >= PHYSIO_RESPONDER_MIN_CSPLUS_TRIALS
    out["physiological_responder_CSR"] = out["n_raw_CSR_ge_0p05"] >= PHYSIO_RESPONDER_MIN_CSPLUS_TRIALS
    return out


def build_simple_differential_flags(acq_df):
    rows = []
    for subject, g in acq_df.groupby("Subject"):
        csminus = g[g["Condition"] == "CS-"]
        csplus = g[g["Domain"] == "CS+"]
        css = g[g["Condition"] == "CSS"]
        csr = g[g["Condition"] == "CSR"]
        rows.append({
            "Subject": subject,
            "simple_diff_CSplus": safe_mean(csplus[SCORE_COL]) - safe_mean(csminus[SCORE_COL]),
            "simple_diff_CSS": safe_mean(css[SCORE_COL]) - safe_mean(csminus[SCORE_COL]),
            "simple_diff_CSR": safe_mean(csr[SCORE_COL]) - safe_mean(csminus[SCORE_COL]),
        })
    return pd.DataFrame(rows)


def _beta_for_subject(g, target_conditions, label):
    d = g[g["Condition"].isin(["CS-"] + list(target_conditions))].copy()
    if d.empty or d["Condition"].nunique() < 2:
        return {f"habituation_beta_{label}": np.nan, f"habituation_beta_trial_{label}": np.nan, f"habituation_n_{label}": int(len(d))}
    y = d[SCORE_COL].astype(float).to_numpy()
    csplus_indicator = d["Condition"].isin(target_conditions).astype(float).to_numpy()
    trial_z = d["Trial_Z"].astype(float).to_numpy()
    X = np.column_stack([np.ones(len(d)), csplus_indicator, trial_z])
    try:
        beta = np.linalg.lstsq(X, y, rcond=None)[0]
    except np.linalg.LinAlgError:
        beta = [np.nan, np.nan, np.nan]
    return {f"habituation_beta_{label}": float(beta[1]), f"habituation_beta_trial_{label}": float(beta[2]), f"habituation_n_{label}": int(len(y))}


def build_habituation_adjusted_flags(acq_df):
    rows = []
    targets = {"CSplus": ["CSS", "CSR"], "CSS": ["CSS"], "CSR": ["CSR"]}
    for subject, g in acq_df.groupby("Subject"):
        row = {"Subject": subject}
        for label, target_conditions in targets.items():
            row.update(_beta_for_subject(g, target_conditions, label))
        rows.append(row)
    return pd.DataFrame(rows)


def build_late_phase_sensitivity_flags(acq_df):
    rows = []
    targets = {"CSplus": ["CSS", "CSR"], "CSS": ["CSS"], "CSR": ["CSR"]}
    for subject, g in acq_df.groupby("Subject"):
        row = {"Subject": subject}
        for label, target_conditions in targets.items():
            target = g[g["Condition"].isin(target_conditions)]
            csminus = g[g["Condition"] == "CS-"]
            early_diff = safe_mean(target[target["Window"] == "Early"][SCORE_COL]) - safe_mean(csminus[csminus["Window"] == "Early"][SCORE_COL])
            late_diff = safe_mean(target[target["Window"] == "Late"][SCORE_COL]) - safe_mean(csminus[csminus["Window"] == "Late"][SCORE_COL])
            row[f"late_phase_diff_early_{label}"] = early_diff
            row[f"late_phase_diff_late_{label}"] = late_diff
            row[f"late_phase_diff_change_{label}"] = late_diff - early_diff
            row[f"late_phase_learner_{label}"] = (late_diff > 0) and ((late_diff - early_diff) > 0)
        rows.append(row)
    return pd.DataFrame(rows)

acq_for_learners = all_scr[all_scr["Phase"] == "Acquisition"].copy()
physio_flags = build_physio_flags(acq_for_learners)
simple_flags = build_simple_differential_flags(acq_for_learners)
habituation_flags = build_habituation_adjusted_flags(acq_for_learners)
late_phase_flags = build_late_phase_sensitivity_flags(acq_for_learners)
learner_df = physio_flags.merge(simple_flags, on="Subject", how="outer").merge(habituation_flags, on="Subject", how="outer").merge(late_phase_flags, on="Subject", how="outer")
for label in ["CSplus", "CSS", "CSR"]:
    learner_df[f"simple_learner_{label}"] = learner_df["physiological_responder_CSplus"] & (learner_df[f"simple_diff_{label}"] > 0)
    learner_df[f"habituation_adjusted_learner_{label}"] = learner_df["physiological_responder_CSplus"] & (learner_df[f"habituation_beta_{label}"] > 0)
    learner_df[f"late_phase_sensitivity_learner_{label}"] = learner_df[f"late_phase_learner_{label}"].astype(bool)
for prefix in ["simple_learner", "habituation_adjusted_learner", "late_phase_sensitivity_learner"]:
    learner_df[f"{prefix}_either_CSS_or_CSR"] = learner_df[[f"{prefix}_CSS", f"{prefix}_CSR"]].any(axis=1)
    learner_df[f"{prefix}_both_CSS_and_CSR"] = learner_df[[f"{prefix}_CSS", f"{prefix}_CSR"]].all(axis=1)
learner_df = learner_df.merge(meta, on="Subject", how="left") if not meta.empty else learner_df

def summarize_definitions(df, by=None):
    definitions = [
        ("Physiological SCR responder", "physiological_responder_CSplus"),
        ("Simple acquisition differential: pooled CS+ > CS-", "simple_learner_CSplus"),
        ("Habituation-adjusted beta: pooled CS+ > CS-", "habituation_adjusted_learner_CSplus"),
        ("Late-phase sensitivity: pooled CS+", "late_phase_sensitivity_learner_CSplus"),
    ]
    rows = []
    if by is None:
        groups = [((), df)]
        by_cols = []
    else:
        by_cols = list(by)
        groups = list(df.groupby(by_cols, dropna=False))
    for key, sub in groups:
        if not isinstance(key, tuple):
            key = (key,)
        base = {col: val for col, val in zip(by_cols, key)}
        denom = sub["Subject"].nunique()
        for definition, col in definitions:
            n = int(sub[col].fillna(False).sum()) if col in sub else 0
            rows.append({**base, "Definition": definition, "N": n, "Total": denom, "Percent_of_total": safe_percent(n, denom)})
    return pd.DataFrame(rows)

learner_summary = summarize_definitions(learner_df)
learner_by_group = summarize_definitions(learner_df, by=["Group"]) if "Group" in learner_df.columns else pd.DataFrame()
learner_by_group_drug = summarize_definitions(learner_df, by=["Group", "Drug"]) if {"Group", "Drug"}.issubset(learner_df.columns) else pd.DataFrame()

print("Overall learner rates")
display(learner_summary)
print("Learner rates by Group")
display(learner_by_group)
print("Learner rates by Group x Drug")
display(learner_by_group_drug)

analysis_sets = {
    "All subjects": set(learner_df["Subject"]),
    "Physiological responders": set(learner_df.loc[learner_df["physiological_responder_CSplus"].fillna(False), "Subject"]),
    "Simple CS+ learners": set(learner_df.loc[learner_df["simple_learner_CSplus"].fillna(False), "Subject"]),
    "Habituation-adjusted CS+ learners": set(learner_df.loc[learner_df["habituation_adjusted_learner_CSplus"].fillna(False), "Subject"]),
    "Late-phase sensitivity CS+ learners": set(learner_df.loc[learner_df["late_phase_sensitivity_learner_CSplus"].fillna(False), "Subject"]),
}

plot_defs = {
    "Physiological SCR responder": "Physiological\nSCR responder",
    "Simple acquisition differential: pooled CS+ > CS-": "Simple\nCS+ > CS-",
    "Habituation-adjusted beta: pooled CS+ > CS-": "Habituation\nadjusted beta",
    "Late-phase sensitivity: pooled CS+": "Late-phase\nsensitivity",
}
plot_df = learner_by_group[learner_by_group["Definition"].isin(plot_defs)].copy()
plot_df["Definition_short"] = plot_df["Definition"].map(plot_defs)
fig, ax = plt.subplots(figsize=(9.5, 5.2))
sns.barplot(data=plot_df, x="Definition_short", y="Percent_of_total", hue="Group", hue_order=GROUP_ORDER, palette=GROUP_PALETTE, ax=ax)
ax.set_title("Responder and learner rates by diagnostic group")
ax.set_ylabel("Percent of group")
ax.set_xlabel("")
ax.set_ylim(0, max(100, plot_df["Percent_of_total"].max() * 1.2 if not plot_df.empty else 100))
clean_axis(ax)
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=2, fontsize=8)
plt.tight_layout()
save_current_fig("scr_responder_learner_rates_by_group.png")
plt.show()

````

## 5. Statistical Helpers For Phase-Level And Dynamic Models


````python
def _subject_metadata_from_group(g):
    return {
        "Group": g["Group"].dropna().iloc[0] if "Group" in g and g["Group"].notna().any() else np.nan,
        "Drug": g["Drug"].dropna().iloc[0] if "Drug" in g and g["Drug"].notna().any() else np.nan,
    }


def subject_phase_metrics(df, phase, subject_set, analysis_set, window="All"):
    sub = df[(df["Phase"] == phase) & (df["Subject"].isin(subject_set))].copy()
    if window != "All":
        sub = sub[sub["Window"] == window]
    rows = []
    for subject, g in sub.groupby("Subject"):
        meta_row = _subject_metadata_from_group(g)
        csminus = safe_mean(g.loc[g["Condition"] == "CS-", SCORE_COL])
        css = safe_mean(g.loc[g["Condition"] == "CSS", SCORE_COL])
        csr = safe_mean(g.loc[g["Condition"] == "CSR", SCORE_COL])
        csplus = safe_mean(g.loc[g["Domain"] == "CS+", SCORE_COL])
        rows.append({
            "AnalysisSet": analysis_set,
            "Subject": subject,
            "Group": meta_row["Group"],
            "Drug": meta_row["Drug"],
            "Phase": phase,
            "Window": window,
            "Mean_CSminus": csminus,
            "Mean_CSS": css,
            "Mean_CSR": csr,
            "Mean_CSplus": csplus,
            "CS+ - CS-": csplus - csminus,
            "CSR - CSS": csr - css,
            "CSS - CS-": css - csminus,
            "CSR - CS-": csr - csminus,
            "(CSR - CS-) - (CSS - CS-)": (csr - csminus) - (css - csminus),
        })
    return pd.DataFrame(rows)


def collect_phase_metrics(df, phase, analysis_sets, windows=("All",)):
    tables = []
    for set_name in ANALYSIS_SET_ORDER:
        if set_name not in analysis_sets:
            continue
        for window in windows:
            tables.append(subject_phase_metrics(df, phase, analysis_sets[set_name], set_name, window=window))
    return pd.concat(tables, ignore_index=True) if tables else pd.DataFrame()


def one_sample_metric_stats(metric_df, metrics, alternative="greater"):
    rows = []
    for analysis_set in metric_df["AnalysisSet"].dropna().unique():
        for phase in metric_df["Phase"].dropna().unique():
            for window in metric_df["Window"].dropna().unique():
                sub = metric_df[(metric_df["AnalysisSet"] == analysis_set) & (metric_df["Phase"] == phase) & (metric_df["Window"] == window)]
                for metric in metrics:
                    for group in GROUP_ORDER:
                        vals = pd.to_numeric(sub.loc[sub["Group"] == group, metric], errors="coerce").dropna()
                        if len(vals) >= 2:
                            tres = stats.ttest_1samp(vals, 0.0, alternative=alternative)
                            d = vals.mean() / vals.std(ddof=1) if vals.std(ddof=1) > 0 else np.nan
                        else:
                            tres = None
                            d = np.nan
                        rows.append({
                            "AnalysisSet": analysis_set,
                            "Phase": phase,
                            "Window": window,
                            "Metric": metric,
                            "Test": f"{group} vs 0",
                            "Group": group,
                            "N": int(len(vals)),
                            "Mean": float(vals.mean()) if len(vals) else np.nan,
                            "SE": float(vals.sem()) if len(vals) > 1 else np.nan,
                            "t": float(tres.statistic) if tres is not None else np.nan,
                            "p": float(tres.pvalue) if tres is not None else np.nan,
                            "EffectSize": float(d) if not pd.isna(d) else np.nan,
                        })
    return pd.DataFrame(rows)


def group_difference_metric_stats(metric_df, metrics):
    rows = []
    for analysis_set in metric_df["AnalysisSet"].dropna().unique():
        for phase in metric_df["Phase"].dropna().unique():
            for window in metric_df["Window"].dropna().unique():
                sub = metric_df[(metric_df["AnalysisSet"] == analysis_set) & (metric_df["Phase"] == phase) & (metric_df["Window"] == window)]
                for metric in metrics:
                    sad = pd.to_numeric(sub.loc[sub["Group"] == "SAD", metric], errors="coerce").dropna()
                    hc = pd.to_numeric(sub.loc[sub["Group"] == "HC", metric], errors="coerce").dropna()
                    if len(sad) >= 2 and len(hc) >= 2:
                        tres = stats.ttest_ind(sad, hc, equal_var=False)
                        pooled_var = ((len(sad) - 1) * sad.var(ddof=1) + (len(hc) - 1) * hc.var(ddof=1)) / (len(sad) + len(hc) - 2)
                        d = (sad.mean() - hc.mean()) / np.sqrt(pooled_var) if pooled_var > 0 else np.nan
                    else:
                        tres = None
                        d = np.nan
                    rows.append({
                        "AnalysisSet": analysis_set,
                        "Phase": phase,
                        "Window": window,
                        "Metric": metric,
                        "Test": "SAD vs HC",
                        "N_SAD": int(len(sad)),
                        "N_HC": int(len(hc)),
                        "Mean_SAD": float(sad.mean()) if len(sad) else np.nan,
                        "Mean_HC": float(hc.mean()) if len(hc) else np.nan,
                        "Mean_Diff_SAD_minus_HC": float(sad.mean() - hc.mean()) if len(sad) and len(hc) else np.nan,
                        "t": float(tres.statistic) if tres is not None else np.nan,
                        "p": float(tres.pvalue) if tres is not None else np.nan,
                        "EffectSize": float(d) if not pd.isna(d) else np.nan,
                    })
    return pd.DataFrame(rows)


def factorial_group_drug_tests(metric_df, metrics):
    rows = []
    for analysis_set in metric_df["AnalysisSet"].dropna().unique():
        for phase in metric_df["Phase"].dropna().unique():
            for window in metric_df["Window"].dropna().unique():
                sub = metric_df[(metric_df["AnalysisSet"] == analysis_set) & (metric_df["Phase"] == phase) & (metric_df["Window"] == window)].copy()
                sub = sub.dropna(subset=["Group", "Drug"])
                for metric in metrics:
                    d = sub[["Subject", "Group", "Drug", metric]].dropna().copy()
                    if d["Group"].nunique() < 2 or d["Drug"].nunique() < 2 or len(d) < 8:
                        rows.append({"AnalysisSet": analysis_set, "Phase": phase, "Window": window, "Metric": metric, "Term": "Group:Drug", "N": len(d), "coef": np.nan, "t": np.nan, "p": np.nan, "note": "insufficient cells"})
                        continue
                    try:
                        model = smf.ols(f"Q('{metric}') ~ C(Group, Treatment(reference='HC')) * C(Drug, Treatment(reference='Placebo'))", data=d).fit()
                        term = "C(Group, Treatment(reference='HC'))[T.SAD]:C(Drug, Treatment(reference='Placebo'))[T.Oxytocin]"
                        rows.append({
                            "AnalysisSet": analysis_set,
                            "Phase": phase,
                            "Window": window,
                            "Metric": metric,
                            "Term": "Group_SAD_x_Drug_Oxytocin",
                            "N": int(model.nobs),
                            "coef": float(model.params.get(term, np.nan)),
                            "t": float(model.tvalues.get(term, np.nan)),
                            "p": float(model.pvalues.get(term, np.nan)),
                            "note": "OLS metric ~ Group * Drug",
                        })
                    except Exception as exc:
                        rows.append({"AnalysisSet": analysis_set, "Phase": phase, "Window": window, "Metric": metric, "Term": "Group:Drug", "N": len(d), "coef": np.nan, "t": np.nan, "p": np.nan, "note": str(exc)})
    return pd.DataFrame(rows)


def long_metric_table(metric_df, metrics):
    return metric_df.melt(id_vars=["AnalysisSet", "Subject", "Group", "Drug", "Phase", "Window"], value_vars=metrics, var_name="Metric", value_name="Score").dropna(subset=["Score"])


def add_metric_sig_labels(ax, stats_df, analysis_set, phase, window):
    sub = stats_df[(stats_df["AnalysisSet"] == analysis_set) & (stats_df["Phase"] == phase) & (stats_df["Window"] == window)].copy()
    if sub.empty:
        return
    xticks = ax.get_xticks()
    labels = [t.get_text() for t in ax.get_xticklabels()]
    y_min, y_max = ax.get_ylim()
    span = y_max - y_min if y_max > y_min else 1
    for _, row in sub.iterrows():
        stars = p_to_stars(row["p"])
        if not stars or row["Metric"] not in labels:
            continue
        x = xticks[labels.index(row["Metric"])]
        y = y_max - span * 0.10
        ax.text(x, y, stars, ha="center", va="bottom", fontsize=13, fontweight="bold", color="black")
        ax.text(x, y - span * 0.055, p_text(row["p"]), ha="center", va="bottom", fontsize=8, color="black")


def _annotate_bar_pvalues(ax, plot_sub, metrics, metric_order, group_order=GROUP_ORDER, alternative="two-sided"):
    if plot_sub.empty:
        return
    y_min, y_max = ax.get_ylim()
    span = y_max - y_min if y_max > y_min else 1.0
    group_offsets = {group: offset for group, offset in zip(group_order, np.linspace(-0.18, 0.18, len(group_order)))}
    top_needed = y_max
    for metric_idx, metric in enumerate(metrics):
        metric_label = metric_display_name(metric)
        for group in group_order:
            vals = pd.to_numeric(plot_sub.loc[(plot_sub["Metric"] == metric) & (plot_sub["Group"] == group), "Score"], errors="coerce").dropna()
            if len(vals) < 2:
                continue
            tres = stats.ttest_1samp(vals, 0.0, alternative=alternative)
            mean_val = float(vals.mean())
            se_val = float(vals.sem()) if len(vals) > 1 else 0.0
            x_pos = metric_order.index(metric_label) + group_offsets.get(group, 0.0)
            if mean_val >= 0:
                y_pos = mean_val + se_val + 0.035 * span
                va = "bottom"
            else:
                y_pos = mean_val - se_val - 0.045 * span
                va = "top"
            top_needed = max(top_needed, y_pos + 0.08 * span)
            ax.text(x_pos, y_pos, f"{p_text(tres.pvalue)}{p_to_stars(tres.pvalue)}", ha="center", va=va, fontsize=7.2, color="#222222", rotation=0)
    if top_needed > y_max:
        ax.set_ylim(y_min, top_needed)


def plot_metric_group_bars(metric_df, group_stats_df, analysis_set, phase, window, metrics, title, filename):
    long_df = long_metric_table(metric_df[(metric_df["AnalysisSet"] == analysis_set) & (metric_df["Phase"] == phase) & (metric_df["Window"] == window)], metrics)
    if long_df.empty:
        print(f"No data for {title}")
        return
    long_df["MetricDisplay"] = long_df["Metric"].map(metric_display_name)
    metric_order = [metric_display_name(m) for m in metrics]
    fig_w = max(6.2, len(metrics) * 1.45 + 2.4)
    drug_order = [d for d in ["Placebo", "Oxytocin"] if d in set(long_df["Drug"].dropna().astype(str))]
    if not drug_order:
        drug_order = [None]

    if phase in ["Extinction", "Reinstatement"] and len(drug_order) > 1:
        fig, axes = plt.subplots(len(drug_order), 1, figsize=(fig_w, 3.25 * len(drug_order)), sharex=True, sharey=True)
        axes = np.atleast_1d(axes)
        for ax, drug in zip(axes, drug_order):
            plot_sub = long_df[long_df["Drug"].astype(str) == drug].copy()
            sns.barplot(
                data=plot_sub, x="MetricDisplay", y="Score", hue="Group",
                order=metric_order, hue_order=GROUP_ORDER, palette=GROUP_PALETTE,
                errorbar="se", capsize=0.10, err_kws={"linewidth": 1.1}, width=0.74, ax=ax
            )
            ax.axhline(0, color="#222222", linewidth=0.9)
            ax.set_title(drug)
            ax.set_ylabel("Subject contrast (sqrt uS)")
            ax.set_xlabel("")
            ax.tick_params(axis="x", rotation=0)
            clean_axis(ax)
            _annotate_bar_pvalues(ax, plot_sub, metrics, metric_order, alternative="two-sided")
            handles, labels = ax.get_legend_handles_labels()
            if ax is axes[0]:
                ax.legend(handles[:2], labels[:2], title="Group", frameon=False, loc="upper right")
            elif ax.get_legend() is not None:
                ax.get_legend().remove()
        fig.suptitle(title, y=1.01, fontsize=12)
        plt.tight_layout()
        save_current_fig(filename)
        plt.show()
        return

    fig, ax = plt.subplots(figsize=(fig_w, 3.9))
    sns.barplot(data=long_df, x="MetricDisplay", y="Score", hue="Group", order=metric_order, hue_order=GROUP_ORDER, palette=GROUP_PALETTE, errorbar="se", capsize=0.10, err_kws={"linewidth": 1.1}, width=0.74, ax=ax)
    ax.axhline(0, color="#222222", linewidth=0.9)
    ax.set_title(title)
    ax.set_ylabel("Subject contrast (sqrt uS)")
    ax.set_xlabel("")
    ax.tick_params(axis="x", rotation=0)
    clean_axis(ax)
    _annotate_bar_pvalues(ax, long_df, metrics, metric_order, alternative="two-sided")
    handles, labels = ax.get_legend_handles_labels()
    ax.legend(handles[:2], labels[:2], title="Group", frameon=False, loc="upper right")
    plt.tight_layout()
    save_current_fig(filename)
    plt.show()


def plot_acquisition_condition_bars(metric_df, within_stats_df, group_stats_df, analysis_set, title, filename, window="All"):
    sub = metric_df[(metric_df["AnalysisSet"] == analysis_set) & (metric_df["Phase"] == "Acquisition") & (metric_df["Window"] == window)].copy()
    if sub.empty:
        print(f"No acquisition condition data for {analysis_set} / {window}")
        return

    condition_palette = {"CS-": CONDITION_PALETTE["CS-"], "CS+": CONDITION_PALETTE["CS+"]}
    means = {
        group: {
            "CS-": safe_mean(sub.loc[sub["Group"] == group, "Mean_CSminus"]),
            "CS+": safe_mean(sub.loc[sub["Group"] == group, "Mean_CSplus"]),
        }
        for group in GROUP_ORDER
    }
    diff_se = {}
    for group in GROUP_ORDER:
        diff_vals = pd.to_numeric(sub.loc[sub["Group"] == group, "CS+ - CS-"], errors="coerce").dropna()
        diff_se[group] = float(diff_vals.sem()) if len(diff_vals) > 1 else 0.0

    fig, ax = plt.subplots(figsize=(4.8, 3.8))
    x = np.arange(len(GROUP_ORDER), dtype=float)
    width = 0.34
    offsets = {"CS-": -width / 2, "CS+": width / 2}
    for condition in ["CS-", "CS+"]:
        heights = [means[group][condition] for group in GROUP_ORDER]
        yerr = [diff_se[group] for group in GROUP_ORDER]
        ax.bar(
            x + offsets[condition], heights, width=width, label=condition,
            color=condition_palette[condition], edgecolor="white", linewidth=0.7,
            yerr=yerr, capsize=4, error_kw={"elinewidth": 1.1, "capthick": 1.1, "ecolor": "#303030"},
        )

    ax.set_xticks(x)
    ax.set_xticklabels(GROUP_ORDER)
    ax.set_title(title)
    ax.set_ylabel("Mean SCR (sqrt uS)")
    ax.set_xlabel("Diagnostic group")
    clean_axis(ax)
    ax.legend(title="Condition", frameon=False, loc="upper right")
    text_rows = []
    stat_sub = within_stats_df[(within_stats_df["AnalysisSet"] == analysis_set) & (within_stats_df["Phase"] == "Acquisition") & (within_stats_df["Window"] == window) & (within_stats_df["Metric"] == "CS+ - CS-")]
    for group in GROUP_ORDER:
        row = stat_sub[stat_sub["Group"] == group]
        if not row.empty:
            text_rows.append(f"{group}: {p_text(row.iloc[0]['p'])}{p_to_stars(row.iloc[0]['p'])}")
    group_row = group_stats_df[(group_stats_df["AnalysisSet"] == analysis_set) & (group_stats_df["Phase"] == "Acquisition") & (group_stats_df["Window"] == window) & (group_stats_df["Metric"] == "CS+ - CS-")]
    if not group_row.empty:
        text_rows.append(f"SAD-HC: {p_text(group_row.iloc[0]['p'])}{p_to_stars(group_row.iloc[0]['p'])}")
    if text_rows:
        ax.text(0.02, 0.98, "\n".join(text_rows), transform=ax.transAxes, ha="left", va="top", fontsize=7.8, bbox={"boxstyle": "round,pad=0.25", "facecolor": "white", "edgecolor": "#BDBDBD", "alpha": 0.88})
    plt.tight_layout()
    save_current_fig(filename)
    plt.show()


def prepare_dynamic_df(df, phase, subject_set, mode):
    sub = df[(df["Phase"] == phase) & (df["Subject"].isin(subject_set))].copy()
    if mode == "acquisition_csplus":
        sub["CS_type"] = np.where(sub["Domain"] == "CS+", "CSplus", "CSminus")
        agg_cols = ["Subject", "Group", "Drug", "Phase", "CS_type", "ConditionTrial"]
        sub = sub.groupby(agg_cols, dropna=False, as_index=False, observed=True).agg({SCORE_COL: "mean", "Trial_Z": "mean"})
    elif mode == "css_csr":
        sub = sub[sub["Condition"].isin(["CSS", "CSR"])].copy()
        sub["CS_type"] = sub["Condition"].astype(str)
    elif mode == "all_conditions":
        sub = sub[sub["Condition"].isin(CONDITION_ORDER)].copy()
        sub["CS_type"] = sub["Condition"].astype(str)
    else:
        sub["CS_type"] = sub["Condition"].astype(str).replace({"CS-": "CSminus"})
    sub["CS_type"] = pd.Categorical(sub["CS_type"].astype(str), categories=sorted(sub["CS_type"].dropna().astype(str).unique()))
    sub = sub.dropna(subset=[SCORE_COL, "Trial_Z", "Group", "Drug", "CS_type", "Subject"])
    return sub


def fit_dynamic_lme(dynamic_df, phase, analysis_set, mode):
    rows = []
    if dynamic_df.empty or dynamic_df["Subject"].nunique() < 4:
        return pd.DataFrame([{"AnalysisSet": analysis_set, "Phase": phase, "Mode": mode, "Term": np.nan, "coef": np.nan, "z_or_t": np.nan, "p": np.nan, "N_trials": len(dynamic_df), "N_subjects": dynamic_df["Subject"].nunique() if not dynamic_df.empty else 0, "note": "insufficient data"}])
    formula = f"{SCORE_COL} ~ C(CS_type) * C(Group) * C(Drug) * Trial_Z"
    try:
        model = smf.mixedlm(formula, data=dynamic_df, groups=dynamic_df["Subject"]).fit(reml=False, method="lbfgs", maxiter=250, disp=False)
        for term, coef in model.params.items():
            rows.append({
                "AnalysisSet": analysis_set,
                "Phase": phase,
                "Mode": mode,
                "Term": term,
                "coef": float(coef),
                "z_or_t": float(model.tvalues.get(term, np.nan)),
                "p": float(model.pvalues.get(term, np.nan)),
                "N_trials": int(model.nobs),
                "N_subjects": int(dynamic_df["Subject"].nunique()),
                "note": "MixedLM sqrt_scr ~ CS_type * Group * Drug * Trial_Z + (1|Subject)",
            })
    except Exception as exc:
        try:
            model = smf.ols(formula, data=dynamic_df).fit(cov_type="cluster", cov_kwds={"groups": dynamic_df["Subject"]})
            for term, coef in model.params.items():
                rows.append({
                    "AnalysisSet": analysis_set,
                    "Phase": phase,
                    "Mode": mode,
                    "Term": term,
                    "coef": float(coef),
                    "z_or_t": float(model.tvalues.get(term, np.nan)),
                    "p": float(model.pvalues.get(term, np.nan)),
                    "N_trials": int(model.nobs),
                    "N_subjects": int(dynamic_df["Subject"].nunique()),
                    "note": f"Cluster-robust OLS fallback after MixedLM failure: {exc}",
                })
        except Exception as exc2:
            rows.append({"AnalysisSet": analysis_set, "Phase": phase, "Mode": mode, "Term": np.nan, "coef": np.nan, "z_or_t": np.nan, "p": np.nan, "N_trials": len(dynamic_df), "N_subjects": dynamic_df["Subject"].nunique(), "note": f"model failed: {exc2}"})
    return pd.DataFrame(rows)


def _annotate_acquisition_trial_stars(ax, sub, alpha=0.05):
    if sub.empty or not {"Subject", "ConditionTrial", "CS_type", SCORE_COL}.issubset(sub.columns):
        return
    y_min, y_max = ax.get_ylim()
    span = y_max - y_min if y_max > y_min else 1.0
    top_needed = y_max
    for trial in range(1, 9):
        trial_df = sub[sub["ConditionTrial"] == trial]
        wide = trial_df.pivot_table(index="Subject", columns="CS_type", values=SCORE_COL, aggfunc="mean")
        if not {"CSplus", "CSminus"}.issubset(wide.columns):
            continue
        diff = (wide["CSplus"] - wide["CSminus"]).dropna()
        if len(diff) < 2:
            continue
        tres = stats.ttest_1samp(diff, 0.0, alternative="greater")
        if np.isfinite(tres.pvalue) and tres.pvalue < alpha:
            means = trial_df.groupby("CS_type")[SCORE_COL].mean()
            ses = trial_df.groupby("CS_type")[SCORE_COL].sem().fillna(0.0)
            y_pos = float((means + ses).max()) + 0.035 * span
            top_needed = max(top_needed, y_pos + 0.065 * span)
            ax.text(trial, y_pos, "*", ha="center", va="bottom", fontsize=12, fontweight="bold", color="#111111", clip_on=False)
    if top_needed > y_max:
        ax.set_ylim(y_min, top_needed)


def plot_dynamic_timecourse(df, phase, subject_set, analysis_set, mode, filename):
    dyn = prepare_dynamic_df(df, phase, subject_set, mode=mode)
    if dyn.empty:
        print(f"No dynamic data for {phase} / {analysis_set} / {mode}")
        return dyn
    hue_order_all = [c for c in ["CS-", "CSS", "CSR", "CSminus", "CSplus"] if c in set(dyn["CS_type"].astype(str))]
    pal = {"CS-": CONDITION_PALETTE["CS-"], "CSS": CONDITION_PALETTE["CSS"], "CSR": CONDITION_PALETTE["CSR"], "CSminus": CONDITION_PALETTE["CS-"], "CSplus": CONDITION_PALETTE["CS+"]}
    if phase == "Acquisition":
        fig, axes = plt.subplots(1, 2, figsize=(7.8, 3.8), sharex=True, sharey=True)
        legend_handles = None
        legend_labels = None
        for ax, group, letter in zip(axes, GROUP_ORDER, ["A", "B"]):
            sub = dyn[dyn["Group"] == group]
            if sub.empty:
                ax.set_title(f"{group}\n(no data)")
                clean_axis(ax)
                panel_label(ax, letter)
                continue
            sns.lineplot(data=sub, x="ConditionTrial", y=SCORE_COL, hue="CS_type", hue_order=hue_order_all, palette=pal, marker="o", linewidth=1.8, markersize=4.5, errorbar="se", err_style="band", ax=ax)
            if legend_handles is None:
                legend_handles, legend_labels = ax.get_legend_handles_labels()
            if ax.get_legend() is not None:
                ax.get_legend().remove()
            ax.set_title(group)
            ax.set_xlabel("Trial within condition")
            ax.set_ylabel("Mean SCR (sqrt uS)" if group == "HC" else "")
            ax.set_xticks(range(1, 9))
            clean_axis(ax)
            _annotate_acquisition_trial_stars(ax, sub)
            panel_label(ax, letter)
        if legend_handles is not None:
            fig.legend(legend_handles, legend_labels, title="Condition", frameon=False, loc="center right", bbox_to_anchor=(1.05, 0.5))
        fig.suptitle(f"{phase} trial-wise SCR dynamics: {analysis_set}", y=1.04, fontsize=12)
        plt.tight_layout(rect=(0, 0, 0.88, 0.98))
        save_current_fig(filename)
        plt.show()
        return dyn
    fig, axes = plt.subplots(2, 2, figsize=(9.2, 6.4), sharex=True, sharey=True)
    panel_order = [("HC", "Placebo"), ("HC", "Oxytocin"), ("SAD", "Placebo"), ("SAD", "Oxytocin")]
    letters = ["A", "B", "C", "D"]
    legend_handles = None
    legend_labels = None
    for ax, (group, drug), letter in zip(axes.ravel(), panel_order, letters):
        sub = dyn[(dyn["Group"] == group) & (dyn["Drug"] == drug)]
        if sub.empty:
            ax.set_title(f"{group} / {drug}\n(no data)")
            clean_axis(ax)
            panel_label(ax, letter)
            continue
        hue_order = [c for c in ["CS-", "CSS", "CSR", "CSminus", "CSplus"] if c in set(sub["CS_type"].astype(str))]
        sns.lineplot(data=sub, x="ConditionTrial", y=SCORE_COL, hue="CS_type", hue_order=hue_order, palette=pal, marker="o", linewidth=1.8, markersize=4.5, errorbar="se", err_style="band", ax=ax)
        if legend_handles is None:
            legend_handles, legend_labels = ax.get_legend_handles_labels()
        if ax.get_legend() is not None:
            ax.get_legend().remove()
        ax.set_title(f"{group} / {drug}")
        ax.set_xlabel("Trial within condition" if ax in axes[1, :] else "")
        ax.set_ylabel("Mean SCR (sqrt uS)" if ax in axes[:, 0] else "")
        ax.set_xticks(range(1, 9))
        clean_axis(ax)
        panel_label(ax, letter)
    if legend_handles is not None:
        fig.legend(legend_handles, legend_labels, title="Condition", frameon=False, loc="center right", bbox_to_anchor=(1.03, 0.5))
    fig.suptitle(f"{phase} trial-wise SCR dynamics: {analysis_set}", y=1.02, fontsize=12)
    plt.tight_layout(rect=(0, 0, 0.90, 0.98))
    save_current_fig(filename)
    plt.show()
    return dyn



````

## 9. Dynamic Mixed-Effects Models And Phase-Level Summary


````python
dynamic_tables = []
for set_name in ANALYSIS_SET_ORDER:
    if set_name not in analysis_sets:
        continue
    dyn_acq = prepare_dynamic_df(all_scr, "Acquisition", analysis_sets[set_name], mode="acquisition_csplus")
    dyn_ext = prepare_dynamic_df(all_scr, "Extinction", analysis_sets[set_name], mode="css_csr")
    dyn_rein = prepare_dynamic_df(all_scr, "Reinstatement", analysis_sets[set_name], mode="css_csr")
    dynamic_tables.append(fit_dynamic_lme(dyn_acq, "Acquisition", set_name, "CSplus_vs_CSminus"))
    dynamic_tables.append(fit_dynamic_lme(dyn_ext, "Extinction", set_name, "CSS_vs_CSR"))
    dynamic_tables.append(fit_dynamic_lme(dyn_rein, "Reinstatement", set_name, "CSS_vs_CSR"))
dynamic_model_stats = pd.concat(dynamic_tables, ignore_index=True)
print("Dynamic mixed-effects / cluster-robust model statistics")
display(dynamic_model_stats)

phase_summary_long = pd.concat([
    long_metric_table(acq_metrics, ["CS+ - CS-"]),
    long_metric_table(ext_metrics, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
    long_metric_table(rein_metrics, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
], ignore_index=True)
phase_summary_all = phase_summary_long[phase_summary_long["Window"] == "All"].copy()

fig, axes = plt.subplots(1, 3, figsize=(15, 4.8), sharey=False)
for ax, phase, metric in zip(axes, ["Acquisition", "Extinction", "Reinstatement"], ["CS+ - CS-", "CSR - CSS", "CSR - CSS"]):
    sub = phase_summary_all[(phase_summary_all["Phase"] == phase) & (phase_summary_all["Metric"] == metric) & (phase_summary_all["AnalysisSet"] == "All subjects")]
    sns.pointplot(data=sub, x="Group", y="Score", hue="Drug", order=GROUP_ORDER, palette={"Placebo": "#6C8EBF", "Oxytocin": "#D98C5F"}, dodge=0.28, errorbar="se", markers="o", linestyles="", ax=ax)
    ax.axhline(0, color="#222222", linewidth=1)
    ax.set_title(f"{phase}\n{metric}")
    ax.set_xlabel("")
    ax.set_ylabel("Subject contrast (sqrt uS)" if phase == "Acquisition" else "")
    clean_axis(ax)
fig.suptitle("Primary phase-level SCR effects by Group and Drug", y=1.05)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_group_drug_summary.png")
plt.show()

PRIMARY_EFFECT_ORDER = [
    "Acquisition\nCS+ - CS-",
    "Extinction\nCSR - CSS",
    "Extinction\nCSS - CS-",
    "Extinction\nCSR - CS-",
    "Reinstatement\nCSR - CSS",
    "Reinstatement\nCSS - CS-",
    "Reinstatement\nCSR - CS-",
]

def _annotate_positive_effect_stars(ax, plot_df, effect_order, analysis_order, alpha=0.05):
    if plot_df.empty:
        return
    y_min, y_max = ax.get_ylim()
    span = y_max - y_min if y_max > y_min else 1.0
    offsets = {label: offset for label, offset in zip(analysis_order, np.linspace(-0.225, 0.225, len(analysis_order)))}
    top_needed = y_max
    for effect_idx, effect in enumerate(effect_order):
        for analysis_set in analysis_order:
            vals = pd.to_numeric(
                plot_df.loc[(plot_df["Effect"] == effect) & (plot_df["AnalysisSet"] == analysis_set), "Score"],
                errors="coerce",
            ).dropna()
            if len(vals) < 2:
                continue
            tres = stats.ttest_1samp(vals, 0.0, alternative="greater")
            if np.isfinite(tres.pvalue) and tres.pvalue < alpha:
                mean_val = float(vals.mean())
                se_val = float(vals.sem()) if len(vals) > 1 else 0.0
                x_pos = effect_idx + offsets.get(analysis_set, 0.0)
                y_pos = mean_val + se_val + 0.025 * span
                top_needed = max(top_needed, y_pos + 0.055 * span)
                ax.text(x_pos, y_pos, "*", ha="center", va="bottom", fontsize=10, fontweight="bold", color="#111111", clip_on=False)
    if top_needed > y_max:
        ax.set_ylim(y_min, top_needed)


def _effect_rows(df, phase, metric, effect_label, sign=1.0):
    d = df[(df["Phase"] == phase) & (df["Metric"] == metric)].copy()
    d["Score"] = sign * d["Score"]
    d["Effect"] = effect_label
    return d

def _build_primary_effect_summary(df):
    summary_parts = []
    summary_parts.append(_effect_rows(df, "Acquisition", "CS+ - CS-", "Acquisition\nCS+ - CS-"))
    summary_parts.append(_effect_rows(df, "Extinction", "CSR - CSS", "Extinction\nCSR - CSS"))
    summary_parts.append(_effect_rows(df, "Extinction", "CSS - CS-", "Extinction\nCSS - CS-"))
    summary_parts.append(_effect_rows(df, "Extinction", "CSR - CS-", "Extinction\nCSR - CS-"))
    summary_parts.append(_effect_rows(df, "Reinstatement", "CSR - CSS", "Reinstatement\nCSR - CSS"))
    summary_parts.append(_effect_rows(df, "Reinstatement", "CSS - CS-", "Reinstatement\nCSS - CS-"))
    summary_parts.append(_effect_rows(df, "Reinstatement", "CSR - CS-", "Reinstatement\nCSR - CS-"))
    return pd.concat(summary_parts, ignore_index=True)

primary_effect_summary_long = _build_primary_effect_summary(phase_summary_all)
primary_effect_window_summary_long = _build_primary_effect_summary(phase_summary_long)

fig, ax = plt.subplots(figsize=(12.5, 4.8))
sns.pointplot(data=primary_effect_summary_long, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
ax.axhline(0, color="#222222", linewidth=1)
ax.set_title("Primary SCR effects across phases and sensitivity sets")
ax.set_ylabel("Subject contrast (sqrt uS)")
ax.set_xlabel("")
ax.tick_params(axis="x", rotation=0)
clean_axis(ax)
ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_analysis_set_summary.png")
plt.show()

fig, axes = plt.subplots(3, 1, figsize=(13.5, 10.0), sharex=True, sharey=True)
for ax, window in zip(axes, ["All", "Early", "Late"]):
    sub = primary_effect_window_summary_long[primary_effect_window_summary_long["Window"] == window].copy()
    sns.pointplot(data=sub, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
    ax.axhline(0, color="#222222", linewidth=1)
    ax.set_title(window)
    ax.set_ylabel("Subject contrast\n(sqrt uS)")
    ax.set_xlabel("")
    clean_axis(ax)
    _annotate_positive_effect_stars(ax, sub, PRIMARY_EFFECT_ORDER, ANALYSIS_SET_ORDER)
    if ax is axes[0]:
        ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
    elif ax.get_legend() is not None:
        ax.get_legend().remove()
axes[-1].tick_params(axis="x", rotation=0)
fig.suptitle("Primary SCR effects by analysis window", y=1.01, fontsize=12)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_analysis_set_by_window.png")
plt.show()

# Sensitivity visualization: exclude the first trial within each phase/condition.
# All = trials 2-8; Early = trials 2-4; Late = trials 5-8.
all_scr_no_first = all_scr[all_scr["ConditionTrial"] >= 2].copy()
all_scr_no_first["Window"] = np.where(
    all_scr_no_first["ConditionTrial"].between(2, 4),
    "Early",
    np.where(all_scr_no_first["ConditionTrial"].between(5, 8), "Late", np.nan),
)
acq_metrics_no_first = collect_phase_metrics(all_scr_no_first, "Acquisition", analysis_sets, windows=("All", "Early", "Late"))
ext_metrics_no_first = collect_phase_metrics(all_scr_no_first, "Extinction", analysis_sets, windows=("All", "Early", "Late"))
rein_metrics_no_first = collect_phase_metrics(all_scr_no_first, "Reinstatement", analysis_sets, windows=("All", "Early", "Late"))
phase_summary_no_first_long = pd.concat([
    long_metric_table(acq_metrics_no_first, ["CS+ - CS-"]),
    long_metric_table(ext_metrics_no_first, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
    long_metric_table(rein_metrics_no_first, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
], ignore_index=True)
primary_effect_no_first_window_summary_long = _build_primary_effect_summary(phase_summary_no_first_long)

fig, axes = plt.subplots(3, 1, figsize=(13.5, 10.0), sharex=True, sharey=True)
for ax, window in zip(axes, ["All", "Early", "Late"]):
    sub = primary_effect_no_first_window_summary_long[primary_effect_no_first_window_summary_long["Window"] == window].copy()
    sns.pointplot(data=sub, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
    ax.axhline(0, color="#222222", linewidth=1)
    title_suffix = {"All": "trials 2-8", "Early": "trials 2-4", "Late": "trials 5-8"}[window]
    ax.set_title(f"{window} ({title_suffix})")
    ax.set_ylabel("Subject contrast\n(sqrt uS)")
    ax.set_xlabel("")
    clean_axis(ax)
    _annotate_positive_effect_stars(ax, sub, PRIMARY_EFFECT_ORDER, ANALYSIS_SET_ORDER)
    if ax is axes[0]:
        ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
    elif ax.get_legend() is not None:
        ax.get_legend().remove()
axes[-1].tick_params(axis="x", rotation=0)
fig.suptitle("Primary SCR effects by analysis window, excluding first trial", y=1.01, fontsize=12)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_analysis_set_by_window_exclude_trial1.png")
plt.show()

# Sliding-window visualization: 3-trial windows moving one trial at a time.
# Windows are condition-specific trials 1-3, 2-4, ..., 6-8.
def _sliding_window_phase_metrics(df, phase, analysis_sets, starts=range(1, 7), width=3):
    tables = []
    for start in starts:
        stop = start + width - 1
        d = df[df["ConditionTrial"].between(start, stop)].copy()
        d["Window"] = f"T{start}-T{stop}"
        tables.append(collect_phase_metrics(d, phase, analysis_sets, windows=(f"T{start}-T{stop}",)))
    return pd.concat(tables, ignore_index=True) if tables else pd.DataFrame()

acq_metrics_sliding3 = _sliding_window_phase_metrics(all_scr, "Acquisition", analysis_sets)
ext_metrics_sliding3 = _sliding_window_phase_metrics(all_scr, "Extinction", analysis_sets)
rein_metrics_sliding3 = _sliding_window_phase_metrics(all_scr, "Reinstatement", analysis_sets)
phase_summary_sliding3_long = pd.concat([
    long_metric_table(acq_metrics_sliding3, ["CS+ - CS-"]),
    long_metric_table(ext_metrics_sliding3, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
    long_metric_table(rein_metrics_sliding3, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
], ignore_index=True)
primary_effect_sliding3_summary_long = _build_primary_effect_summary(phase_summary_sliding3_long)
SLIDING3_WINDOW_ORDER = [f"T{i}-T{i+2}" for i in range(1, 7)]

fig, axes = plt.subplots(len(SLIDING3_WINDOW_ORDER), 1, figsize=(13.5, 17.5), sharex=True, sharey=True)
for ax, window in zip(axes, SLIDING3_WINDOW_ORDER):
    sub = primary_effect_sliding3_summary_long[primary_effect_sliding3_summary_long["Window"] == window].copy()
    sns.pointplot(data=sub, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
    ax.axhline(0, color="#222222", linewidth=1)
    ax.set_title(window)
    ax.set_ylabel("Subject contrast\n(sqrt uS)")
    ax.set_xlabel("")
    clean_axis(ax)
    _annotate_positive_effect_stars(ax, sub, PRIMARY_EFFECT_ORDER, ANALYSIS_SET_ORDER)
    if ax is axes[0]:
        ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
    elif ax.get_legend() is not None:
        ax.get_legend().remove()
axes[-1].tick_params(axis="x", rotation=0)
fig.suptitle("Primary SCR effects in 3-trial sliding windows", y=1.005, fontsize=12)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_analysis_set_sliding3.png")
plt.show()

def _plot_sliding_summary_by_group(summary_df, window_order, group, title, filename, fig_height, drug=None):
    plot_df = summary_df[summary_df["Group"] == group].copy()
    if drug is not None:
        plot_df = plot_df[plot_df["Drug"].astype(str) == drug].copy()
    if plot_df.empty:
        label = f"{group} / {drug}" if drug is not None else group
        print(f"No sliding-window data for {label}: {title}")
        return
    fig, axes = plt.subplots(len(window_order), 1, figsize=(13.5, fig_height), sharex=True, sharey=True)
    axes = np.atleast_1d(axes)
    for ax, window in zip(axes, window_order):
        sub = plot_df[plot_df["Window"] == window].copy()
        sns.pointplot(data=sub, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
        ax.axhline(0, color="#222222", linewidth=1)
        ax.set_title(window)
        ax.set_ylabel("Subject contrast\n(sqrt uS)")
        ax.set_xlabel("")
        clean_axis(ax)
        _annotate_positive_effect_stars(ax, sub, PRIMARY_EFFECT_ORDER, ANALYSIS_SET_ORDER)
        if ax is axes[0]:
            ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
        elif ax.get_legend() is not None:
            ax.get_legend().remove()
    axes[-1].tick_params(axis="x", rotation=0)
    fig.suptitle(title, y=1.005, fontsize=12)
    plt.tight_layout()
    save_current_fig(filename)
    plt.show()



def _annotate_positive_window_stars(ax, plot_df, window_order, analysis_order, alpha=0.05):
    """Mark uncorrected one-sample effects where subject contrasts are greater than zero."""
    if plot_df.empty:
        return
    y_min, y_max = ax.get_ylim()
    span = y_max - y_min if y_max > y_min else 1.0
    offsets = {label: offset for label, offset in zip(analysis_order, np.linspace(-0.225, 0.225, len(analysis_order)))}
    top_needed = y_max
    for window_idx, window in enumerate(window_order):
        for analysis_set in analysis_order:
            vals = pd.to_numeric(
                plot_df.loc[(plot_df["Window"] == window) & (plot_df["AnalysisSet"] == analysis_set), "Score"],
                errors="coerce",
            ).dropna()
            if len(vals) < 2:
                continue
            tres = stats.ttest_1samp(vals, 0.0, alternative="greater")
            if np.isfinite(tres.pvalue) and tres.pvalue < alpha:
                mean_val = float(vals.mean())
                se_val = float(vals.sem()) if len(vals) > 1 else 0.0
                x_pos = window_idx + offsets.get(analysis_set, 0.0)
                y_pos = mean_val + se_val + 0.025 * span
                top_needed = max(top_needed, y_pos + 0.055 * span)
                ax.text(x_pos, y_pos, "*", ha="center", va="bottom", fontsize=10, fontweight="bold", color="#111111", clip_on=False)
    if top_needed > y_max:
        ax.set_ylim(y_min, top_needed)


def _plot_sliding_summary_by_effect(summary_df, window_order, title, filename, fig_height, group=None, drug=None):
    """Plot one phase effect per row, with all sliding windows on the x-axis."""
    plot_df = summary_df.copy()
    if group is not None:
        plot_df = plot_df[plot_df["Group"] == group].copy()
    if drug is not None:
        plot_df = plot_df[plot_df["Drug"].astype(str) == drug].copy()
    if plot_df.empty:
        label_parts = [x for x in [group, drug] if x is not None]
        label = " / ".join(label_parts) if label_parts else "combined"
        print(f"No effect-row sliding-window data for {label}: {title}")
        return
    fig, axes = plt.subplots(len(PRIMARY_EFFECT_ORDER), 1, figsize=(13.5, fig_height), sharex=True, sharey=True)
    axes = np.atleast_1d(axes)
    for ax, effect in zip(axes, PRIMARY_EFFECT_ORDER):
        sub = plot_df[plot_df["Effect"] == effect].copy()
        sns.pointplot(
            data=sub,
            x="Window",
            y="Score",
            hue="AnalysisSet",
            order=window_order,
            hue_order=ANALYSIS_SET_ORDER,
            palette=SET_PALETTE,
            dodge=0.45,
            errorbar="se",
            markers="o",
            linestyles="",
            ax=ax,
        )
        ax.axhline(0, color="#222222", linewidth=1)
        ax.set_title(effect.replace("\n", " / "))
        ax.set_ylabel("Subject contrast\n(sqrt uS)")
        ax.set_xlabel("")
        clean_axis(ax)
        _annotate_positive_window_stars(ax, sub, window_order, ANALYSIS_SET_ORDER)
        if ax is axes[0]:
            ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
        elif ax.get_legend() is not None:
            ax.get_legend().remove()
    axes[-1].set_xlabel("Sliding window")
    axes[-1].tick_params(axis="x", rotation=0)
    fig.suptitle(title, y=1.005, fontsize=12)
    plt.tight_layout()
    save_current_fig(filename)
    plt.show()

for group in GROUP_ORDER:
    _plot_sliding_summary_by_group(
        primary_effect_sliding3_summary_long, SLIDING3_WINDOW_ORDER, group,
        f"Primary SCR effects in 3-trial sliding windows: {group}",
        f"scr_primary_phase_effects_analysis_set_sliding3_{group.lower()}.png",
        fig_height=17.5,
    )

for group in GROUP_ORDER:
    for drug in ["Placebo", "Oxytocin"]:
        _plot_sliding_summary_by_group(
            primary_effect_sliding3_summary_long, SLIDING3_WINDOW_ORDER, group,
            f"Primary SCR effects in 3-trial sliding windows: {group} / {drug}",
            f"scr_primary_phase_effects_analysis_set_sliding3_{group.lower()}_{drug.lower()}.png",
            fig_height=17.5,
            drug=drug,
        )


_plot_sliding_summary_by_effect(
    primary_effect_sliding3_summary_long,
    SLIDING3_WINDOW_ORDER,
    "Primary SCR effects in 3-trial sliding windows, effect-row layout",
    "scr_primary_phase_effects_by_effect_sliding3.png",
    fig_height=20.0,
)

for group in GROUP_ORDER:
    _plot_sliding_summary_by_effect(
        primary_effect_sliding3_summary_long,
        SLIDING3_WINDOW_ORDER,
        f"Primary SCR effects in 3-trial sliding windows, effect-row layout: {group}",
        f"scr_primary_phase_effects_by_effect_sliding3_{group.lower()}.png",
        fig_height=20.0,
        group=group,
    )

for group in GROUP_ORDER:
    for drug in ["Placebo", "Oxytocin"]:
        _plot_sliding_summary_by_effect(
            primary_effect_sliding3_summary_long,
            SLIDING3_WINDOW_ORDER,
            f"Primary SCR effects in 3-trial sliding windows, effect-row layout: {group} / {drug}",
            f"scr_primary_phase_effects_by_effect_sliding3_{group.lower()}_{drug.lower()}.png",
            fig_height=20.0,
            group=group,
            drug=drug,
        )

# Sliding-window visualization: 2-trial windows moving one trial at a time.
# Windows are condition-specific trials 1-2, 2-3, ..., 7-8.
acq_metrics_sliding2 = _sliding_window_phase_metrics(all_scr, "Acquisition", analysis_sets, starts=range(1, 8), width=2)
ext_metrics_sliding2 = _sliding_window_phase_metrics(all_scr, "Extinction", analysis_sets, starts=range(1, 8), width=2)
rein_metrics_sliding2 = _sliding_window_phase_metrics(all_scr, "Reinstatement", analysis_sets, starts=range(1, 8), width=2)
phase_summary_sliding2_long = pd.concat([
    long_metric_table(acq_metrics_sliding2, ["CS+ - CS-"]),
    long_metric_table(ext_metrics_sliding2, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
    long_metric_table(rein_metrics_sliding2, ["CSR - CSS", "CSS - CS-", "CSR - CS-"]),
], ignore_index=True)
primary_effect_sliding2_summary_long = _build_primary_effect_summary(phase_summary_sliding2_long)
SLIDING2_WINDOW_ORDER = [f"T{i}-T{i+1}" for i in range(1, 8)]

fig, axes = plt.subplots(len(SLIDING2_WINDOW_ORDER), 1, figsize=(13.5, 20.0), sharex=True, sharey=True)
for ax, window in zip(axes, SLIDING2_WINDOW_ORDER):
    sub = primary_effect_sliding2_summary_long[primary_effect_sliding2_summary_long["Window"] == window].copy()
    sns.pointplot(data=sub, x="Effect", y="Score", hue="AnalysisSet", order=PRIMARY_EFFECT_ORDER, hue_order=ANALYSIS_SET_ORDER, palette=SET_PALETTE, dodge=0.45, errorbar="se", markers="o", linestyles="", ax=ax)
    ax.axhline(0, color="#222222", linewidth=1)
    ax.set_title(window)
    ax.set_ylabel("Subject contrast\n(sqrt uS)")
    ax.set_xlabel("")
    clean_axis(ax)
    _annotate_positive_effect_stars(ax, sub, PRIMARY_EFFECT_ORDER, ANALYSIS_SET_ORDER)
    if ax is axes[0]:
        ax.legend(title="Analysis set", bbox_to_anchor=(1.01, 1), loc="upper left", frameon=False)
    elif ax.get_legend() is not None:
        ax.get_legend().remove()
axes[-1].tick_params(axis="x", rotation=0)
fig.suptitle("Primary SCR effects in 2-trial sliding windows", y=1.005, fontsize=12)
plt.tight_layout()
save_current_fig("scr_primary_phase_effects_analysis_set_sliding2.png")
plt.show()

for group in GROUP_ORDER:
    _plot_sliding_summary_by_group(
        primary_effect_sliding2_summary_long, SLIDING2_WINDOW_ORDER, group,
        f"Primary SCR effects in 2-trial sliding windows: {group}",
        f"scr_primary_phase_effects_analysis_set_sliding2_{group.lower()}.png",
        fig_height=20.0,
    )

for group in GROUP_ORDER:
    for drug in ["Placebo", "Oxytocin"]:
        _plot_sliding_summary_by_group(
            primary_effect_sliding2_summary_long, SLIDING2_WINDOW_ORDER, group,
            f"Primary SCR effects in 2-trial sliding windows: {group} / {drug}",
            f"scr_primary_phase_effects_analysis_set_sliding2_{group.lower()}_{drug.lower()}.png",
            fig_height=20.0,
            drug=drug,
        )


_plot_sliding_summary_by_effect(
    primary_effect_sliding2_summary_long,
    SLIDING2_WINDOW_ORDER,
    "Primary SCR effects in 2-trial sliding windows, effect-row layout",
    "scr_primary_phase_effects_by_effect_sliding2.png",
    fig_height=20.0,
)

for group in GROUP_ORDER:
    _plot_sliding_summary_by_effect(
        primary_effect_sliding2_summary_long,
        SLIDING2_WINDOW_ORDER,
        f"Primary SCR effects in 2-trial sliding windows, effect-row layout: {group}",
        f"scr_primary_phase_effects_by_effect_sliding2_{group.lower()}.png",
        fig_height=20.0,
        group=group,
    )

for group in GROUP_ORDER:
    for drug in ["Placebo", "Oxytocin"]:
        _plot_sliding_summary_by_effect(
            primary_effect_sliding2_summary_long,
            SLIDING2_WINDOW_ORDER,
            f"Primary SCR effects in 2-trial sliding windows, effect-row layout: {group} / {drug}",
            f"scr_primary_phase_effects_by_effect_sliding2_{group.lower()}_{drug.lower()}.png",
            fig_height=20.0,
            group=group,
            drug=drug,
        )
````
